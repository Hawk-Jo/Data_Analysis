# 기본 데이터셋 설정

```python
import os
import json
import pandas as pd
from sqlalchemy import create_engine
```

```python
# 1. JSON 파일 읽기 경로 설정
folder_path = "./src/additional_hackle_1"
files = [f for f in os.listdir(folder_path) if f.endswith('.json')]
```

```python
# MySQL 연결 설정
from pw import pw
engine = create_engine(f"mysql+pymysql://root:{pw}@localhost:3306/final")
```

```python
# 2. 전체 파일에서 가능한 모든 컬럼 수집
all_columns = set()

for file in files:
    file_path = os.path.join(folder_path, file)
    with open(file_path, 'r') as f:
        data = json.load(f)
    df = pd.json_normalize(data)
    all_columns.update(df.columns)
```

```python
# 3. 각 JSON 파일 처리 및 적재
for file in files:
    file_path = os.path.join(folder_path, file)
    with open(file_path, 'r') as f:
        data = json.load(f)

    # JSON 데이터 평탄화
    df = pd.json_normalize(data)

    # 누락된 컬럼 채우기
    for col in all_columns:
        if col not in df.columns:
            df[col] = None

    # 파일명으로 테이블 이름 생성
    table_name = file.replace('.json', '')

    # 데이터 MySQL에 적재
    df.to_sql(table_name, con=engine, if_exists='replace', index=False)

    print(f"파일 {file}이 테이블 {table_name}로 적재되었습니다.")
```

## 출석 체크 기반 리텐션 확인

```python
from ast import literal_eval
attendance['attendance_date_list'] = attendance['attendance_date_list'].apply(lambda x: literal_eval(x) if type(x) == str else [])

exploded_df = attendance.explode("attendance_date_list")

exploded_df["attendance_date_list"] = pd.to_datetime(exploded_df["attendance_date_list"])
exploded_df["first_attendance_date"] = exploded_df.groupby("user_id")["attendance_date_list"].transform("min")

# Step 0: 날짜 차이 계산
exploded_df["days_since_first"] = (exploded_df["attendance_date_list"] - exploded_df["first_attendance_date"]).dt.days

# Step 1: Cohort 정의 (첫 출석일을 기준으로 코호트 설정)
exploded_df["cohort"] = exploded_df["first_attendance_date"].dt.to_period("M")

# Step 2: 특정 리텐션 기간 계산 (D+1, D+7, D+30)
# 사용자별 첫 출석일을 기준으로 일별 재방문 여부를 계산
retention_df = exploded_df.groupby(["cohort", "days_since_first"])["user_id"].nunique().reset_index()
retention_df = retention_df.pivot(index="cohort", columns="days_since_first", values="user_id")

# Step 3: 리텐션 퍼센트 계산 (각 코호트의 첫날 사용자 수 기준)
retention_df = retention_df.div(retention_df[0], axis=0) * 100  # 첫날 유저 수로 나눠 백분율 계산
```

```python
retention_df.to_csv('retention_df.csv',index=False, encoding= 'utf-8-sig')
exploded_df.to_csv('exploded_df.csv',index=False, encoding= 'utf-8-sig')
```

```python
retention_df_32_days = retention_df.loc[:, :32]

# 코호트 분석 표 시각화
plt.figure(figsize=(18, 6))

# Heatmap으로 표시 (Retention %)
sns.heatmap(
    retention_df_32_days,
    annot=True,
    fmt=".1f",
    cmap="Blues",
    cbar_kws={"label": "Retention Rate (%)"},
    linewidths=0.5
)

plt.title("Cohort Retention Analysis", fontsize=16)
plt.xlabel("Days Since First Attendance", fontsize=12)
plt.ylabel("Cohort (First Attendance Month)", fontsize=12)
plt.tight_layout()

# 그래프 출력
plt.show()
```

```python
# 슈퍼 유저, 스태프 제거
user = user[(user['is_superuser'] == 0) & (user['is_staff'] == 0)]

# 친구 수, 차단 유저 수, 숨김 유저 수 계산
user['friend_id_list'] = user['friend_id_list'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith('[') else [])
user['block_user_id_list'] = user['block_user_id_list'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith('[') else [])
user['hide_user_id_list'] = user['hide_user_id_list'].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith('[') else [])

user['friend_count'] = user['friend_id_list'].apply(len)
user['block_count'] = user['block_user_id_list'].apply(len)
user['hide_count'] = user['hide_user_id_list'].apply(len)

user.drop(['is_superuser', 'is_staff'], axis=1, inplace=True)
user
```

```python
# 유저 성별 매핑 준비
user_gen = user[['user_id','gender','friend_id_list']]
user_gen_map = user_gen[['user_id','gender']].set_index('user_id')['gender'].to_dict()
```

```python
import ast

# 이성 친구 수를 계산하는 함수 수정
def calculate_opposite_gender_count_and_ratio(row):
    user_gender = row["gender"]
    friend_ids = row["friend_id_list"]

    # 문자열로 저장된 경우 리스트로 변환
    if isinstance(friend_ids, str):
        try:
            friend_ids = ast.literal_eval(friend_ids)  # 안전한 변환 사용
        except (SyntaxError, ValueError):
            friend_ids = []

    # friend_ids가 NaN이거나 리스트가 아닐 경우 빈 리스트 처리
    if not isinstance(friend_ids, list):
        friend_ids = []

    # 이성 친구 수 계산
    opposite_gender_count = sum(1 for fid in friend_ids if user_gen_map.get(fid) and user_gen_map[fid] != user_gender)
    total_friends = len(friend_ids)

    # 비율 계산 (친구가 없을 경우 0으로 설정)
    opposite_gender_ratio = opposite_gender_count / total_friends if total_friends > 0 else 0

    return pd.Series([opposite_gender_count, opposite_gender_ratio])

# 이성 친구 수 및 비율 계산 적용
user_gen[["opposite_gender_count", "opposite_gender_ratio"]] = user_gen.apply(
    calculate_opposite_gender_count_and_ratio, axis=1
)

# 결과 확인
user_gen

```

```python
opposite_gender_friend = user_gen[['user_id','opposite_gender_count','opposite_gender_ratio']]
opposite_gender_friend.rename(columns={'user_id':'sender_id','opposite_gender_count':'opposite_gender_friend_count','opposite_gender_ratio':'opposite_gender_friend_ratio'}, inplace=True)
opposite_gender_friend.to_csv('opposite_gender_friend.csv', encoding='utf-8-sig')
```

```python
# 유저별 활동 데이터 계산

# 유저별 마지막 출석일 계산
merged_user_attendance['last_attendance_date'] = merged_user_attendance['attendance_date_list'].apply(
    lambda x: pd.to_datetime(eval(x))[-1] if isinstance(x, str) and len(eval(x)) > 0 else pd.NaT
)

# 활동 지속시간(가입일과 마지막 활동일 간의 차이) 산출
merged_user_attendance['created_at'] = pd.to_datetime(merged_user_attendance['created_at'])  # object를 datetime으로 변환
merged_user_attendance['activity_duration'] = (
    # last_attendance_date에 시간 정보가 없어 created_at도 시간 정보 제거하고 계산
    merged_user_attendance['last_attendance_date'] - merged_user_attendance['created_at'].dt.normalize()
).dt.days

# last_attendance_date가 NaN이 아닌 경우 최소 1로 설정
merged_user_attendance.loc[
    merged_user_attendance['last_attendance_date'].notna() & (merged_user_attendance['activity_duration'] < 1),
    'activity_duration'
] = 1

# 결과 확인
merged_user_attendance[['user_id', 'created_at', 'last_attendance_date', 'activity_duration']]
```

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import MinMaxScaler

# 데이터 준비
X = user_q_info[['duration_days','count']]  # 실제 데이터에 맞게 수정 필요
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# DBSCAN 파라미터 설정
min_samples = 5  # 일반적으로 5~10 사이의 값을 사용

# 최근접 이웃(k-NN) 찾기 (k = min_samples)
neigh = NearestNeighbors(n_neighbors=min_samples)
neigh.fit(X_scaled)
distances, indices = neigh.kneighbors(X_scaled)

# 거리값 정렬 (각 포인트에서 k번째 이웃까지의 거리)
k_distances = np.sort(distances[:, -1])  # 마지막 열(min_samples번째 거리) 기준 정렬

# 그래프 그리기
plt.plot(k_distances)
plt.xlabel("Points sorted by distance")
plt.ylabel(f"{min_samples}-th Nearest Neighbor Distance")
plt.title("K-Distance Graph")
plt.show()
```

### 클러스터링 - GMM

```python
# 클러스터링 - AIC, BIC로 최적의 GMM 클러스터 수 찾기

from sklearn.mixture import GaussianMixture

X = user_q_info[['duration_days','count']]
aic_values = []
bic_values = []

for n in range(1, 11):  # 1부터 시작 가능
    gmm = GaussianMixture(n_components=n, random_state=42)
    gmm.fit(X)
    aic_values.append(gmm.aic(X))
    bic_values.append(gmm.bic(X))

plt.plot(range(1, 11), aic_values, label="AIC", marker="o")
plt.plot(range(1, 11), bic_values, label="BIC", marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("AIC / BIC Score")
plt.legend()
plt.title("AIC and BIC for GMM")
plt.show()
```

# 2. 친구 수별 리텐션 분석 코드

```python
# 친구 수별 리텐션 분석
path = "/content/drive/MyDrive/Colab Notebooks/pjdata/votes/"
path_hackle = "/content/drive/MyDrive/Colab Notebooks/pjdata/hackle/"

# 데이터 불러오기
df = pd.read_csv(path_hackle + 'hackle_final(0718-0821).csv')
df_accounts_user = pd.read_csv(path + "accounts_user.csv")

# friend_count 컬럼 생성
def count_friends(friend_id_list):
    if isinstance(friend_id_list, str):  # 문자열 형태인 경우
        friend_list = eval(friend_id_list) # 문자열을 리스트로 변환
        return len(friend_list)
    elif isinstance(friend_id_list, list): # 리스트 형태인 경우
        return len(friend_id_list)
    else: # 결측값 또는 다른 형태인 경우
        return 0

# 함수 반영 및 추가된 데이터 병합
df_accounts_user['friend_count'] = df_accounts_user['friend_id_list'].apply(count_friends)
df_accounts_user = df_accounts_user[['id', 'friend_count']]
df = df.merge(df_accounts_user, left_on='user_id', right_on='id', how='left')

# 친구 수 그룹화 함수
def categorize_friend_count(friends):
    if friends <= 4:
        return "Very-Low"
    elif 5 <= friends <= 10:
        return "Low"
    elif 11 <= friends <= 40:
        return "Mid-Low"
    elif 41 <= friends <= 70:
        return "Mid-High"
    elif 71 <= friends <= 100:
        return "High"
    else:
        return "Very-High"

# 함수 반영
df["friend_count_group_2"] = df["friend_count"].apply(categorize_friend_count)


# click_question_start 액션에 대한 리텐션 데이터
# conversion window = 30 days, click_question_click
# 날짜 데이터 변환
df['event_datetime'] = pd.to_datetime(df['event_datetime'])

# click_question_start 이벤트만 필터링
start_df = df[df['event_key'] == 'click_question_start'].copy()

# 30일 이내 2번 이상 발생한 user_id 찾기
start_df = start_df.sort_values(['user_id', 'event_datetime'])
start_df['time_diff'] = start_df.groupby('user_id')['event_datetime'].diff().dt.days

# 30일 내에 두 번째 이벤트가 있는 user_id 필터링
start_valid_users = start_df.groupby('user_id').filter(lambda x: (x['time_diff'] <= 30).sum() > 0)['user_id'].unique()

# 원본 데이터에서 해당 user_id의 로그만 유지
start_filtered_df = df[df['user_id'].isin(start_valid_users)].copy()

# 날짜 데이터 변환
start_filtered_df['event_datetime'] = pd.to_datetime(start_filtered_df['event_datetime'])

# 'click_question_start' 이벤트만 필터링하고, user_id별로 정렬
df_event_sorted = start_filtered_df[start_filtered_df['event_key'] == 'click_question_start'].sort_values(by=['user_id', 'event_datetime'])

# 각 user_id별 두 번째 'click_question_start' 발생 시점 찾기
df_second_event = df_event_sorted.groupby('user_id').nth(1).reset_index()
df_second_event = df_second_event[['user_id', 'event_datetime']].rename(columns={'event_datetime': 'first_event_time'})

# 원본 데이터에 'first_event_time' 병합
df_filtered = start_filtered_df.merge(df_second_event, on='user_id', how='left')

# 'days_since_first' 계산
df_filtered['days_since_first'] = (df_filtered['event_datetime'] - df_filtered['first_event_time']).dt.days

df_filtered[['user_id', 'event_key', 'event_datetime', 'first_event_time', 'days_since_first', 'friend_count_group']]


# click_question_open 액션에 대한 리텐션 데이터
# conversion window = 30 days, click_question_open
# 날짜 데이터 변환
df['event_datetime'] = pd.to_datetime(df['event_datetime'])

# click_question_open 이벤트만 필터링
open_df = df[df['event_key'] == 'click_question_open'].copy()

# 30일 이내 2번 이상 발생한 user_id 찾기
open_df = open_df.sort_values(['user_id', 'event_datetime'])
open_df['time_diff'] = open_df.groupby('user_id')['event_datetime'].diff().dt.days

# 30일 내에 두 번째 이벤트가 있는 user_id 필터링
open_valid_users = open_df.groupby('user_id').filter(lambda x: (x['time_diff'] <= 30).sum() > 0)['user_id'].unique()

# 원본 데이터에서 해당 user_id의 로그만 유지
open_filtered_df = df[df['user_id'].isin(open_valid_users)].copy()

# 날짜 데이터 변환
open_filtered_df['event_datetime'] = pd.to_datetime(open_filtered_df['event_datetime'])

# 'click_question_open' 이벤트만 필터링하고, user_id별로 정렬
df_event_sorted = open_filtered_df[open_filtered_df['event_key'] == 'click_question_open'].sort_values(by=['user_id', 'event_datetime'])

# 각 user_id별 두 번째 'click_question_start' 발생 시점 찾기
df_second_event = df_event_sorted.groupby('user_id').nth(1).reset_index()
df_second_event = df_second_event[['user_id', 'event_datetime']].rename(columns={'event_datetime': 'first_event_time'})

# 원본 데이터에 'first_event_time' 병합
df_filtered = open_filtered_df.merge(df_second_event, on='user_id', how='left')

# 'days_since_first' 계산
df_filtered['days_since_first'] = (df_filtered['event_datetime'] - df_filtered['first_event_time']).dt.days

df_filtered[['user_id', 'event_key', 'event_datetime', 'first_event_time', 'days_since_first', 'friend_count_group']]


# 리텐션율 계산 및 시각화
# 리텐션율 계산 함수
def calculate_retention(df, brackets, title):
    retention_rates = {}

    # 각 그룹별 리텐션 계산
    for friend_group in df["friend_count_group"].unique():
        df_filtered = df[df["friend_count_group"] == friend_group]

        # user_id, days_since_first 조합이 유일한 데이터 유지
        df_filtered = df_filtered.drop_duplicates(subset=["user_id", "days_since_first"])

        # 초기 유저 수
        initial_users = df_filtered["user_id"].nunique()

        # 리텐션율 계산
        retention_rates[friend_group] = []
        for bracket, (start, end) in brackets.items():
            retained_users = df_filtered[(df_filtered["days_since_first"] >= start) &
                                         (df_filtered["days_since_first"] <= end)]["user_id"].nunique()
            retention_rate = (retained_users / initial_users) * 100 if initial_users > 0 else 0
            retention_rates[friend_group].append(retention_rate)

    # 전체 유저의 리텐션율 계산
    df_total = df.drop_duplicates(subset=["user_id", "days_since_first"])
    initial_users_total = df_total["user_id"].nunique()

    total_retention_rates = []
    for bracket, (start, end) in brackets.items():
        retained_users_total = df_total[(df_total["days_since_first"] >= start) &
                                        (df_total["days_since_first"] <= end)]["user_id"].nunique()
        total_retention_rate = (retained_users_total / initial_users_total) * 100 if initial_users_total > 0 else 0
        total_retention_rates.append(total_retention_rate)

    # 그래프 생성
    plt.figure(figsize=(8, 5))

    # friend_group별 리텐션 그래프
    for friend_group, rates in retention_rates.items():
        plt.plot(list(brackets.keys()), rates, marker='o', label=friend_group)

    # 전체 유저 리텐션율 (회색 점선 추가)
    plt.plot(list(brackets.keys()), total_retention_rates, marker='o', linestyle='--', color='gray', label="전체 유저")

    plt.xlabel("Days Bracket")
    plt.ylabel("Retention Rate (%)")
    plt.title(f"Retention Rate by Friend Count Group with Overall Trend ({title})")
    plt.legend(title="Friend Count Group")
    plt.ylim(0, 100)
    plt.grid(True)
    plt.show()

# Bracket 설정
brackets_start = {
    "Day 0": (0, 0),
    "Day 1-7": (1, 7),
    "Day 8-14": (8, 14),
}

brackets_open = {
    "Day 0": (0, 0),
    "Day 1-9": (1, 9),
    "Day 10-18": (10, 18),
}

# df_question_start 리텐션 계산
calculate_retention(df_question_start, brackets_start, "click_question_start")

# df_question_open 리텐션 계산
calculate_retention(df_question_open, brackets_open, "click_question_open")
```

```python
# 친구 수별 리텐션 분석
path = "/content/drive/MyDrive/Colab Notebooks/pjdata/votes/"
path_hackle = "/content/drive/MyDrive/Colab Notebooks/pjdata/hackle/"

# 데이터 불러오기
df = pd.read_csv(path_hackle + 'hackle_final(0718-0821).csv')
df_accounts_user = pd.read_csv(path + "accounts_user.csv")

# 불러온 데이터 가공
df_hackle_user = df_hackle.copy()
df_hackle_user = df_hackle_user[['user_id']]
df_hackle_user.drop_duplicates(subset=['user_id'])
df_hackle_user = df_hackle_user['user_id'].unique()

df_accounts_user_hackle = df_accounts_user[df_accounts_user['id'].isin(df_hackle_user)]
df_accounts_user_hackle.rename(columns={'id': 'user_id'}, inplace=True)
df_accounts_user_hackle = df_accounts_user_hackle[['user_id', 'gender', 'friend_id_list']]

df_accounts_user_hackle = df_accounts_user_hackle[['user_id', 'gender']]
df_hackle = df_hackle.merge(df_accounts_user_hackle, on='user_id', how='left')
df_hackle.dropna(inplace=True)

df = df_hakle.copy()


# click_question_start 액션에 대한 리텐션 데이터
# conversion window = 30 days, click_question_click
# 날짜 데이터 변환
df['event_datetime'] = pd.to_datetime(df['event_datetime'])

# click_question_start 이벤트만 필터링
start_df = df[df['event_key'] == 'click_question_start'].copy()

# 30일 이내 2번 이상 발생한 user_id 찾기
start_df = start_df.sort_values(['user_id', 'event_datetime'])
start_df['time_diff'] = start_df.groupby('user_id')['event_datetime'].diff().dt.days

# 30일 내에 두 번째 이벤트가 있는 user_id 필터링
start_valid_users = start_df.groupby('user_id').filter(lambda x: (x['time_diff'] <= 30).sum() > 0)['user_id'].unique()

# 원본 데이터에서 해당 user_id의 로그만 유지
start_filtered_df = df[df['user_id'].isin(start_valid_users)].copy()

# 날짜 데이터 변환
start_filtered_df['event_datetime'] = pd.to_datetime(start_filtered_df['event_datetime'])

# 'click_question_start' 이벤트만 필터링하고, user_id별로 정렬
df_event_sorted = start_filtered_df[start_filtered_df['event_key'] == 'click_question_start'].sort_values(by=['user_id', 'event_datetime'])

# 각 user_id별 두 번째 'click_question_start' 발생 시점 찾기
df_second_event = df_event_sorted.groupby('user_id').nth(1).reset_index()
df_second_event = df_second_event[['user_id', 'event_datetime']].rename(columns={'event_datetime': 'first_event_time'})

# 원본 데이터에 'first_event_time' 병합
df_filtered = start_filtered_df.merge(df_second_event, on='user_id', how='left')

# 'days_since_first' 계산
df_filtered['days_since_first'] = (df_filtered['event_datetime'] - df_filtered['first_event_time']).dt.days

df_filtered[['user_id', 'event_key', 'event_datetime', 'first_event_time', 'days_since_first', 'gender']]

# click_question_open 액션에 대한 리텐션 데이터
# conversion window = 30 days, click_question_open
# 날짜 데이터 변환
# conversion window = 60 days, click_question_open
# 날짜 데이터 변환
df['event_datetime'] = pd.to_datetime(df['event_datetime'])

# click_question_open 이벤트만 필터링
open_df = df[df['event_key'] == 'click_question_open'].copy()

# 30일 이내 2번 이상 발생한 user_id 찾기
open_df = open_df.sort_values(['user_id', 'event_datetime'])
open_df['time_diff'] = open_df.groupby('user_id')['event_datetime'].diff().dt.days

# 30일 내에 두 번째 이벤트가 있는 user_id 필터링
open_valid_users = open_df.groupby('user_id').filter(lambda x: (x['time_diff'] <= 30).sum() > 0)['user_id'].unique()

# 원본 데이터에서 해당 user_id의 로그만 유지
open_filtered_df = df[df['user_id'].isin(open_valid_users)].copy()

# 날짜 데이터 변환
open_filtered_df['event_datetime'] = pd.to_datetime(open_filtered_df['event_datetime'])

# 'click_question_open' 이벤트만 필터링하고, user_id별로 정렬
df_event_sorted = open_filtered_df[open_filtered_df['event_key'] == 'click_question_open'].sort_values(by=['user_id', 'event_datetime'])

# 각 user_id별 두 번째 'click_question_start' 발생 시점 찾기
df_second_event = df_event_sorted.groupby('user_id').nth(1).reset_index()
df_second_event = df_second_event[['user_id', 'event_datetime']].rename(columns={'event_datetime': 'first_event_time'})

# 원본 데이터에 'first_event_time' 병합
df_filtered = open_filtered_df.merge(df_second_event, on='user_id', how='left')

# 'days_since_first' 계산
df_filtered['days_since_first'] = (df_filtered['event_datetime'] - df_filtered['first_event_time']).dt.days

df_filtered[['user_id', 'event_key', 'event_datetime', 'first_event_time', 'days_since_first', 'gender']]


# 언더샘플링
from imblearn.under_sampling import RandomUnderSampler

# 성별 데이터를 숫자로 변환 (M → 0, F → 1)
df_question_start["gender_encoded"] = df_question_start["gender"].map({"M": 0, "F": 1})

# 유저 ID 단위로 다운샘플링을 수행하기 위해 유니크한 유저만 유지
df_unique_users = df_question_start[["user_id", "gender_encoded"]].drop_duplicates()

# 다운샘플링 수행
rus = RandomUnderSampler(random_state=42)  # 랜덤 시드 고정
user_ids_resampled, _ = rus.fit_resample(df_unique_users[["user_id"]], df_unique_users["gender_encoded"])

# 다운샘플링된 유저 ID만 필터링하여 새로운 데이터프레임 생성
df_question_start_sampled = df_question_start[df_question_start["user_id"].isin(user_ids_resampled["user_id"])]

# 결과 확인
print(f"남학생 수: {df_question_start_sampled[df_question_start_sampled['gender'] == 'M']['user_id'].nunique()}")
print(f"여학생 수: {df_question_start_sampled[df_question_start_sampled['gender'] == 'F']['user_id'].nunique()}")

# 샘플링된 데이터 확인
df_question_start_sampled.head()

from imblearn.under_sampling import RandomUnderSampler

# 성별 데이터를 숫자로 변환 (M → 0, F → 1)
df_question_open["gender_encoded"] = df_question_open["gender"].map({"M": 0, "F": 1})

# 유저 ID 단위로 다운샘플링을 수행하기 위해 유니크한 유저만 유지
df_unique_users = df_question_open[["user_id", "gender_encoded"]].drop_duplicates()

# 다운샘플링 수행
rus = RandomUnderSampler(random_state=42)  # 랜덤 시드 고정
user_ids_resampled, _ = rus.fit_resample(df_unique_users[["user_id"]], df_unique_users["gender_encoded"])

# 다운샘플링된 유저 ID만 필터링하여 새로운 데이터프레임 생성
df_question_open_sampled = df_question_open[df_question_open["user_id"].isin(user_ids_resampled["user_id"])]

# 결과 확인
print(f"남학생 수: {df_question_open_sampled[df_question_open_sampled['gender'] == 'M']['user_id'].nunique()}")
print(f"여학생 수: {df_question_open_sampled[df_question_open_sampled['gender'] == 'F']['user_id'].nunique()}")

# 샘플링된 데이터 확인
df_question_open_sampled.head()


# 리텐션율 계산 및 시각화
# 리텐션율 계산 함수
import pandas as pd
import matplotlib.pyplot as plt

# 리텐션율 계산 함수
def calculate_retention(df, brackets, title):
    retention_rates = {}

    # gender 칼럼을 기준으로 그룹화 (남학생 M, 여학생 F)
    unique_genders = df["gender"].dropna().unique()

    # 각 gender별 리텐션 계산
    for gender in unique_genders:
        df_filtered = df[df["gender"] == gender]

        # user_id, days_since_first 조합이 유일한 데이터 유지
        df_filtered = df_filtered.drop_duplicates(subset=["user_id", "days_since_first"])

        # 초기 유저 수
        initial_users = df_filtered["user_id"].nunique()

        # 리텐션율 계산
        retention_rates[gender] = []
        for bracket, (start, end) in brackets.items():
            retained_users = df_filtered[(df_filtered["days_since_first"] >= start) &
                                         (df_filtered["days_since_first"] <= end)]["user_id"].nunique()
            retention_rate = (retained_users / initial_users) * 100 if initial_users > 0 else 0
            retention_rates[gender].append(retention_rate)

    # 전체 유저의 리텐션율 계산
    df_total = df.drop_duplicates(subset=["user_id", "days_since_first"])
    initial_users_total = df_total["user_id"].nunique()

    total_retention_rates = []
    for bracket, (start, end) in brackets.items():
        retained_users_total = df_total[(df_total["days_since_first"] >= start) &
                                        (df_total["days_since_first"] <= end)]["user_id"].nunique()
        total_retention_rate = (retained_users_total / initial_users_total) * 100 if initial_users_total > 0 else 0
        total_retention_rates.append(total_retention_rate)

    # 그래프 생성
    plt.figure(figsize=(8, 5))

    # gender별 리텐션 그래프
    for gender, rates in retention_rates.items():
        label = "남학생" if gender == "M" else "여학생"
        plt.plot(list(brackets.keys()), rates, marker='o', label=label)

    # 전체 유저 리텐션율 (회색 점선 추가)
    plt.plot(list(brackets.keys()), total_retention_rates, marker='o', linestyle='--', color='gray', label="전체 유저")

    plt.xlabel("Days Bracket")
    plt.ylabel("Retention Rate (%)")
    plt.title(f"Retention Rate by Gender with Overall Trend ({title})")
    plt.legend(title="Gender")
    plt.ylim(0, 100)
    plt.grid(True)
    plt.show()

# Bracket 설정
brackets_start = {
    "Day 0": (0, 0),
    "Day 1-7": (1, 7),
    "Day 8-14": (8, 14),
}

brackets_open = {
    "Day 0": (0, 0),
    "Day 1-9": (1, 9),
    "Day 10-18": (10, 18),
}

# df_question_start 리텐션 계산
calculate_retention(df_question_start, brackets_start, "click_question_start")

# df_question_open 리텐션 계산
calculate_retention(df_question_open, brackets_open, "click_question_open")
```

```python
import pandas as pd
import re
import numpy as np
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.decomposition import LatentDirichletAllocation
from collections import Counter

# NLTK 불용어 다운로드 (영어, 한글 불용어 리스트 추가 가능)
nltk.download('stopwords')
from nltk.corpus import stopwords

# 불용어 리스트 (추가 가능)
stopwords_set = set(stopwords.words('english'))
korean_stopwords = {"같은", "싶은", "가장", "제일", "사람", "친구", "사람은", "있는", "있을", "친구는"}  # 추가 가능

# 데이터 로드 (CSV 파일에서 질문 불러오기)
file_path = "pjdata/votes/polls_question.csv"  # 파일 경로 설정
questions_df = pd.read_csv(file_path)

# 데이터 전처리: 질문 텍스트만 추출
questions = questions_df["question_text"].dropna().tolist()

# 텍스트 정제 함수
def clean_text(text):
    text = re.sub(r"[^가-힣a-zA-Z0-9\s]", "", text)  # 특수문자 제거
    text = text.lower()  # 소문자로 변환
    words = text.split()
    words = [word for word in words if word not in stopwords_set and word not in korean_stopwords]
    return " ".join(words)

# 전처리 적용
cleaned_questions = [clean_text(q) for q in questions]

# TF-IDF 벡터화 (불용어 필터링 적용)
vectorizer = TfidfVectorizer(max_features=1000)
tfidf_matrix = vectorizer.fit_transform(cleaned_questions)

# K-Means 클러스터링
num_clusters = 10  # 클러스터 개수 조정 가능
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
kmeans.fit(tfidf_matrix)
cluster_labels = kmeans.labels_

# LDA 토픽 모델링
lda = LatentDirichletAllocation(n_components=num_clusters, random_state=42)
lda.fit(tfidf_matrix)
topic_labels = np.argmax(lda.transform(tfidf_matrix), axis=1)

# 클러스터별 주요 키워드 추출 (K-Means 기준)
top_n = 5  # 상위 키워드 개수 조정 가능
terms = vectorizer.get_feature_names_out()
cluster_keywords = {}

for i in range(num_clusters):
    cluster_center = kmeans.cluster_centers_[i]
    sorted_idx = np.argsort(cluster_center)[::-1][:top_n]
    cluster_keywords[i] = [terms[idx] for idx in sorted_idx]

# 클러스터별 대표 질문 선정
def get_representative_questions(cluster_num, df, num_examples=5):
    return df[df["cluster_label"] == cluster_num]["question"].head(num_examples).tolist()

# 결과 정리
topic_mapping = {}
for cluster in range(num_clusters):
    topic_mapping[cluster] = {
        "keywords": cluster_keywords[cluster],
        "sample_questions": get_representative_questions(cluster, pd.DataFrame({"question": questions, "cluster_label": cluster_labels}))
    }

# 최적의 라벨링 목록 출력
for cluster, info in topic_mapping.items():
    print(f"\n📌 클러스터 {cluster}: 주요 키워드 {info['keywords']}")
    print("대표 질문 예시:")
    for q in info["sample_questions"]:
        print(f"  - {q}")

# CSV 파일로 저장 (라벨링 결과)
labeling_results = pd.DataFrame({
    "question": questions,
    "cluster_label": cluster_labels,
    "topic_label": topic_labels
})
labeling_results.to_csv("labeled_questions.csv", index=False)
print("\n✅ 라벨링 완료! labeled_questions.csv 파일로 저장됨.")
```

### 차이 검정

```python
# 정규성 검정

from scipy.stats import kstest, norm


# 평균과 표준편차를 사용한 정규성 검정
for i in id_added['to_opposite'].unique():
    stat, p = kstest(id_added[id_added['to_opposite'] == i]['opposite_gender_ratio'], 'norm', args=(np.mean(id_added[id_added['to_opposite'] == i]['opposite_gender_ratio']), np.std(id_added[id_added['to_opposite'] == i]['opposite_gender_ratio'])))
    print(f"이성 투표 여부(0 or 1): {i}")
    print(f"Kolmogorov-Smirnov Test: stat={stat}, p-value={p}")

    if p > 0.05:
        print("데이터는 정규성을 따를 가능성이 높음")
    else:
        print("데이터는 정규성을 따르지 않을 가능성이 높음")
```

```python
import scipy.stats as stats


# Mann-Whitney U Test
u_test = stats.mannwhitneyu(id_added[id_added['to_opposite'] == 0]['opposite_gender_ratio'], id_added[id_added['to_opposite'] == 1]['opposite_gender_ratio'])
print(f"Mann-Whitney U Test: U={u_test.statistic}, p-value={u_test.pvalue}")
```

```python
# 투표자의 이성친구 비율과 초성확인의 관계 확인
sender_opposite_friend = fullset.groupby('sender_id')[['opposite_gender_friend_ratio','opposite_gender_ratio']].mean().reset_index()
friend_i_ratio = user_i_exp.merge(sender_opposite_friend, on = 'sender_id', how='left')
friend_i_ratio = friend_i_ratio.merge(user_question_send_count, on = 'sender_id', how = 'left')
friend_i_ratio
```

```python
from scipy.stats import mannwhitneyu
# 통계 검정
group1 = friend_i_ratio[friend_i_ratio['i_exp'] == True]['i_exp']
group2 = friend_i_ratio[friend_i_ratio['i_exp'] == False]['i_exp']
# Mann-Whitney U 검정 실행
stat, p = mannwhitneyu(group1, group2, alternative='two-sided')

print(f"Mann-Whitney U Test: stat={stat}, p-value={p}")

if p > 0.05:
    print("두 그룹 간 유의미한 차이가 없음")
else:
    print("두 그룹 간 유의미한 차이가 있음")
```
